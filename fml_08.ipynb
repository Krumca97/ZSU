# Take a look at this [article](https://builtin.com/data-science/precision-and-recall) and [wiki](https://en.wikipedia.org/wiki/Precision_and_recall) about precision and recall
* 💡 Beware the fact that as we increase precision, we decrease recall and vice-versa.

* **Precision** is the ratio of correctly predicted positive observations to the total predicted positive observations.
    * The question that precision answer is of all passengers that labeled as survived, how many actually survived?
    * High precision relates to the low false positive rate.


* **Recall** is the ratio of correctly predicted positive observations to the all observations in actual class - yes.
    * The question recall answers is: Of all the passengers that truly survived, how many did we label?

* **F1 Score** is the harmonic mean of Precision and Recall. Therefore, this score takes both false positives and false negatives into account.
    * Intuitively it is not as easy to understand as accuracy, but F1 is usually more useful than accuracy, especially if you have an uneven class distribution.
    * 💡 **F1 works best if false positives and false negatives have similar cost**
        * If the cost of false positives and false negatives are very different, it’s better to look at both Precision and Recall.

$ConfMatrix = \begin{bmatrix}
TP & FN\\
FP & TN
\end{bmatrix}$

## Let's take a look at an example:
$|1| = 10$

$|0| = 90$

$M = \begin{bmatrix}
1 & 9\\
0 & 90
\end{bmatrix}$

$Accuracy = \frac{TP+TN}{TP+TN+FP+FN} = \frac{1+90}{1+90+0+9} = \frac{91}{100} = 0.91$

$Precision = \frac{TP}{TP+FP} = \frac{1}{1+0} = 1$

$Recall = \frac{TP}{TP+FN} = \frac{1}{1+9} = \frac{1}{10} = 0.1$

$F1-Score = 2 \cdot \frac{Precision \cdot Recall}{Precision+Recall} = 2 \cdot \frac{1 \cdot 0.1}{1+0.1} = 2 \cdot \frac{0.1}{1.1} = 0.09$

```python
import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier
from sklearn.model_selection import KFold
from sklearn.metrics import f1_score, accuracy_score, confusion_matrix
from sklearn.preprocessing import OrdinalEncoder

#Load dataset
df = pd.read_csv("https://raw.githubusercontent.com/rasvob/VSB-FEI-Fundamentals-of-Machine-Learning-Exercises/master/datasets/titanic.csv",index_col=0)

# Drop features
df = df.drop(["Name", "Ticket"], axis=1)

# Process Cabin (extract deck)
df["Cabin"] = df["Cabin"].astype(str).apply(lambda x: x[0] if x != "nan" else "U")
df["Cabin"] = df["Cabin"].replace({"T": "A"})

# Handle Age missing values
df["Age"] = df["Age"].fillna(df["Age"].median())

# Drop rows with missing Embarked
df = df.dropna()

# Encoding
# Cabin ordinal
cabin_categories = ['A','B','C','D','E','F','G','U']
enc_cabin = OrdinalEncoder(categories=[cabin_categories])
df["Cabin"] = enc_cabin.fit_transform(df[["Cabin"]])

# Sex ordinal
sex_categories = ["male", "female"]
enc_sex = OrdinalEncoder(categories=[sex_categories])
df["Sex"] = enc_sex.fit_transform(df[["Sex"]])

# Embarked one-hot
df = pd.concat([df, pd.get_dummies(df["Embarked"], prefix="Embarked")], axis=1)
df = df.drop("Embarked", axis=1)

# Split into X / y
X = df.drop("Survived", axis=1)
y = df["Survived"]

# 7) DEFAULT RandomForest
rf_default = RandomForestClassifier(random_state=13)

skf = KFold(n_splits=5)
scores_default = []

for train_idx, test_idx in skf.split(X, y):
    X_train, X_test = X.iloc[train_idx], X.iloc[test_idx]
    y_train, y_test = y.iloc[train_idx], y.iloc[test_idx]

    rf_default.fit(X_train, y_train)
    y_pred = rf_default.predict(X_test)
    scores_default.append(f1_score(y_test, y_pred))

print("DEFAULT RandomForest")
print("F1 scores:", scores_default)
print("Mean F1:", np.mean(scores_default))
print()

# Tuning n_estimators
scores_global = []

for n in range(10, 201, 10):
    skf = KFold(n_splits=5)
    scores = []

    for train_idx, test_idx in skf.split(X, y):
        X_train, X_test = X.iloc[train_idx], X.iloc[test_idx]
        y_train, y_test = y.iloc[train_idx], y.iloc[test_idx]

        clf = RandomForestClassifier(n_estimators=n, random_state=13)
        clf.fit(X_train, y_train)
        y_pred = clf.predict(X_test)
        scores.append(f1_score(y_test, y_pred))

    scores_global.append((n, np.mean(scores)))

df_res = pd.DataFrame(scores_global, columns=["n_estimators", "f1"])

print("Best 5 settings:")
print(df_res.sort_values("f1", ascending=False).head())

# Use best model
best_n = int(df_res.sort_values("f1", ascending=False).iloc[0]["n_estimators"])

scores_tuned = []

for train_idx, test_idx in skf.split(X, y):
    X_train, X_test = X.iloc[train_idx], X.iloc[test_idx]
    y_train, y_test = y.iloc[train_idx], y.iloc[test_idx]

    clf = RandomForestClassifier(n_estimators=best_n, random_state=13)
    clf.fit(X_train, y_train)
    y_pred = clf.predict(X_test)
    scores_tuned.append(f1_score(y_test, y_pred))

print("\nTUNED RandomForest")
print("Best n_estimators:", best_n)
print("F1 scores:", scores_tuned)
print("Mean F1:", np.mean(scores_tuned))
print()

# Plot result
sns.lineplot(data=df_res, x="n_estimators", y="f1")
plt.title("F1 score vs n_estimators")
plt.show()

# Markdown summary (printed)
summary = """
##Model: RandomForestClassifier – Summary

Pro experimet jsem si vybral model **RandomForestClassifier**, protože jde o ensemble více rozhodovacích stromů.
Model má stabilnější chování a menší riziko přeučení než samotný DecisionTree.

###Laděné parametry
- `n_estimators` – počet stromů v lese

Ladil jsem hodnoty v rozsahu 10–200.

### Výsledky
- F1 score defaultního modelu: přibližně {:.3f}
- F1 score po naladění: přibližně {:.3f}

Tuned model překonal výkon původního DecisionTreeClassifier i defaultního RandomForest.

###Insight
Z experimentu je vidět, že zvýšení počtu stromů ve RandomForest zlepšilo generalizaci a stabilitu výsledků.
Hyperparameter tuning měl pozitivní vliv a ukázal, že správné nastavení `n_estimators`
může významně zvýšit kvalitu modelu.
"""
```
